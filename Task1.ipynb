### Задача 1.8
> В конкурсе по $5$ номинациям участвуют $10$ кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

* Рассмотрим случай **а)** – призы различные

Комбинации отличаются как составом фильмов, так и их порядком по номинациям, причём один и тот же фильм может получить призы по нескольким номинациям. Это **размещения с повторениями** из $10$ элементов по $5$:

$\overline{A}_{10}^5 = n^k = 10^5$

```python
10**5
```

* Рассмотрим случай **б)** – призы одинаковые

Порядок следования фильмов в комбинации $5$ призёров значения не имеет, повторения допускаются. Это **сочетания с повторениями** из $10$ элементов по $5$:

$\overline{C}_{10}^5 = C_{10+5-1}^5 = C_{14}^5$

```python
int(sc.comb(14, 5))
```

### Задача 1.9
> Сколько существует семизначных чисел, состоящих из цифр $4$, $5$ и $6$, в которых цифра $4$ повторяется $3$ раза, а цифры $5$ и $6$ – по $2$ раза?

Каждое семизначное число отличается от другого только порядком следования цифр, причём $n_1 = 3$, $n_2 = 2$, $n_3 = 2$, а их сумма равна $7$. Это **перестановки с повторениями** из $7$ элементов:

$P_7(3, 2, 2) = \frac{7!}{3! \cdot 2! \cdot 2!}$

```python
int(
    sc.factorial(7)
    / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2))
)
```

## Тема 1.2. Непосредственное вычисление вероятностей
---

### Задача 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребёнок берёт карточки в случайном порядке и прикладывает одну к другой: а) $3$ карточки; б) все $6$ карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* Рассмотрим событие $A$ – получилось слово **ТОР**

Комбинации трёх букв из шести отличаются как составом, так и порядком следования, т.е. являются **размещениями**: $n = A_6^3$. Событию $A$ благоприятствует $m = 1$ случай.

$P(A) = \frac{m}{n} = \frac{1}{A_6^3} = \frac{1}{6 \cdot 5 \cdot 4}$

```python
round(1 / sc.perm(6, 3), 4)
```

* Рассмотрим событие $B$ – получилось слово **ТЕОРИЯ**

Комбинации шести букв из шести отличаются только порядком следования, т.е. являются **перестановками**: $n = P_6 = 6!$. Событию $B$ благоприятствует $m = 1$ случай.

$P(B) = \frac{m}{n} = \frac{1}{P_6} = \frac{1}{6!}$

```python
round(1 / sc.factorial(6), 4)
```

### Задача 1.11
> Используя условие задачи 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

* Рассмотрим событие $B$ – получилось слово **АНАНАС**

Общее число случаев, как и в п. б) задачи 1.10, равно $n = P_6 = 6!$. Но теперь число благоприятствующих случаев больше: перестановка трёх букв А ($P_3 = 3!$ способами) и перестановка двух букв Н ($P_2 = 2!$ способами) не меняет собранное слово. По правилу произведения $m = P_3 \cdot P_2$.

$P(B) = \frac{m}{n} = \frac{P_3 \cdot P_2}{P_6} = \frac{3! \cdot 2!}{6!}$

```python
round(sc.factorial(3) * sc.factorial(2) / sc.factorial(6), 4)
```

Ту же задачу можно решить через **перестановки с повторениями**, из которых событию $B$ благоприятствует одна комбинация:

$P(B) = 1 : P_6(3, 2, 1) = 1 : \frac{6!}{3! \cdot 2! \cdot 1!}$

```python
round(
    1
    / (sc.factorial(6) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(1))),
    4,
)
```

### Задача 1.16
> За круглым столом рассаживаются $5$ мужчин и $5$ женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют $5$ супружеских пар.

* Рассмотрим событие $A$ – никакие два лица одного пола не сядут рядом

Общее число способов рассадки $10$ лиц на $10$ местах: $n = P_{10} = 10!$. Если женщины займут чётные места ($5!$ способами), то мужчины занимают нечётные места (тоже $5!$ способами), и наоборот, поэтому $m_1 = 2 \cdot (5!)^2$.

$P(A) = \frac{m_1}{n} = \frac{2 \cdot (5!)^2}{10!}$

```python
round(2 * sc.factorial(5) ** 2 / sc.factorial(10), 5)
```

* Рассмотрим событие $B$ – супруги, образующие пять пар, сядут рядом

Число благоприятствующих случаев определяется числом перестановок $5$ супружеских пар ($5!$), причём в каждой паре возможна перестановка мужа и жены. По правилу произведения $m_2 = 5! \cdot 2^5$.

$P(B) = \frac{m_2}{n} = \frac{5! \cdot 2^5}{10!}$

```python
round(sc.factorial(5) * 2**5 / sc.factorial(10), 5)
```

### Задача 1.17
> В купейный вагон ($9$ купе по $4$ места) семи пассажирам продано $7$ билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

Общее число способов выбора $7$ любых мест из имеющихся в вагоне $36$:

$n = C_{36}^7$

```python
n = sc.comb(36, 7)
int(n)
```

* Рассмотрим событие $A$ – пассажиры попали в два купе

$2$ купе из $9$ можно выбрать $C_9^2$ способами, а $7$ мест из имеющихся в двух купе $8$ мест – $C_8^7$ способами. По правилу произведения $m_1 = C_9^2 \cdot C_8^7$.

$P(A) = \frac{m_1}{n} = \frac{C_9^2 \cdot C_8^7}{C_{36}^7}$

```python
round(sc.comb(9, 2) * sc.comb(8, 7) / n, 7)
```

* Рассмотрим событие $B$ – пассажиры попали в семь купе

$7$ купе из $9$ можно выбрать $C_9^7$ способами. Семь мест в семи купе получаются выбором по одному месту из четырёх в каждом купе, что возможно $4^7$ способами. Тогда $m_2 = C_9^7 \cdot 4^7$.

$P(B) = \frac{m_2}{n} = \frac{C_9^7 \cdot 4^7}{C_{36}^7}$

```python
round(sc.comb(9, 7) * 4**7 / n, 5)
```

* Рассмотрим событие $D$ – пассажиры попали в три купе

$3$ купе из $9$ можно выбрать $C_9^3$ способами. Возможные варианты распределения $7$ мест по трём купе: $4+2+1$, $3+3+1$, $3+2+2$, а за счёт перестановок купе таких вариантов будет соответственно $6$, $3$ и $3$. Тогда

$m_3 = C_9^3 \left( 6 \cdot C_4^4 C_4^2 C_4^1 + 3 \cdot C_4^3 C_4^3 C_4^1 + 3 \cdot C_4^3 C_4^2 C_4^2 \right)$

$P(D) = \frac{m_3}{n}$

```python
m3 = sc.comb(9, 3) * (
    6 * sc.comb(4, 4) * sc.comb(4, 2) * sc.comb(4, 1)
    + 3 * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(4, 1)
    + 3 * sc.comb(4, 3) * sc.comb(4, 2) * sc.comb(4, 2)
)
round(m3 / n, 5)
```
