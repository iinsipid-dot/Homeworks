### Задача 1.38
> Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от $1$ до $5$)?

* Рассмотрим событие $A$ – тома стоят в порядке нумерации

Различные расстановки пяти томов отличаются только порядком следования, т.е. являются **перестановками**: общее число случаев $n = P_5 = 5!$. Событию $A$ благоприятствует $m = 1$ случай (единственный верный порядок).

$P(A) = \frac{m}{n} = \frac{1}{P_5} = \frac{1}{5!}$

```python
round(1 / sc.factorial(5), 4)
```

### Задача 1.43
> Наудачу взятый телефонный номер состоит из $5$ цифр. Какова вероятность того, что в нём все цифры: а) различные; б) одинаковые; в) нечётные? Известно, что номер телефона не начинается с цифры ноль.

**Общее число случаев.** На первом месте может стоять любая из $9$ цифр ($1, 2, \ldots, 9$), на каждом из остальных четырёх мест – любая из $10$ цифр. По правилу произведения:

$n = 9 \cdot 10^4 = 90\,000$

```python
n = 9 * 10**4
n
```

* Рассмотрим событие $A$ – все цифры номера различны

Первую цифру выбираем $9$ способами (без нуля). Оставшиеся четыре цифры выбираем из $9$ неиспользованных с учётом порядка, т.е. $A_9^4$ способами. По правилу произведения:

$m_1 = 9 \cdot A_9^4 = 9 \cdot 9 \cdot 8 \cdot 7 \cdot 6 = 27\,216$

$P(A) = \frac{m_1}{n} = \frac{9 \cdot A_9^4}{9 \cdot 10^4}$

```python
m1 = 9 * sc.perm(9, 4)
print(int(m1))
round(m1 / n, 4)
```

* Рассмотрим событие $B$ – все цифры номера одинаковы

Подходят только номера вида $11111, 22222, \ldots, 99999$. Номер $00000$ исключён условием, поэтому $m_2 = 9$.

$P(B) = \frac{m_2}{n} = \frac{9}{90\,000}$

```python
round(9 / n, 4)
```

* Рассмотрим событие $D$ – все цифры номера нечётные

Нечётных цифр пять: $1, 3, 5, 7, 9$. Ни одна из них не равна нулю, поэтому ограничение на первую цифру выполняется автоматически, и на каждое из $5$ мест есть $5$ вариантов – это **размещения с повторениями**:

$m_3 = \overline{A}_5^5 = 5^5 = 3125$

$P(D) = \frac{m_3}{n} = \frac{5^5}{90\,000}$

```python
round(5**5 / n, 4)
```

**Проверка полным перебором.** Переберём все пятизначные числа и посчитаем доли напрямую – результат должен совпасть с формулами.

```python
nums = [str(x) for x in range(10000, 100000)]

a = sum(1 for s in nums if len(set(s)) == 5)   # все цифры различны
b = sum(1 for s in nums if len(set(s)) == 1)   # все цифры одинаковы
d = sum(1 for s in nums if set(s) <= set("13579"))  # все цифры нечётны

print(f"а) {a} / {len(nums)} = {round(a / len(nums), 4)}")
print(f"б) {b} / {len(nums)} = {round(b / len(nums), 4)}")
print(f"в) {d} / {len(nums)} = {round(d / len(nums), 4)}")
```

> **Замечание.** В ответах учебника к п. а) указано $1/27216 = 0{,}0000367$. Это противоречит самому способу подсчёта в п. б) и в), где ответ получен как отношение *числа благоприятствующих случаев к общему числу* ($9/90\,000$ и $5^5/90\,000$). Число $27\,216$ – это количество благоприятствующих номеров, поэтому вероятность равна $27\,216 / 90\,000 = 0{,}3024$, что подтверждается и прямым перебором выше.

### Задача 1.48
> В старинной игре в кости необходимо было для выигрыша получить при бросании трёх игральных костей сумму очков, превосходящую $10$. Найти вероятности: а) выпадения $11$ очков; б) выигрыша.

**Общее число случаев.** Каждая из трёх костей независимо даёт одно из $6$ значений – **размещения с повторениями**:

$n = \overline{A}_6^3 = 6^3 = 216$

```python
n = 6**3
n
```

* Рассмотрим событие $A$ – выпало $11$ очков

Сумму $11$ дают следующие наборы очков (в скобках – число перестановок каждого набора):

$(1,4,6) - 3! = 6$; $(2,3,6) - 3! = 6$; $(2,4,5) - 3! = 6$;

$(1,5,5) - \frac{3!}{2!} = 3$; $(3,3,5) - \frac{3!}{2!} = 3$; $(3,4,4) - \frac{3!}{2!} = 3$.

$m_1 = 3 \cdot 6 + 3 \cdot 3 = 27$

$P(A) = \frac{m_1}{n} = \frac{27}{216}$

```python
round(27 / n, 4)
```

* Рассмотрим событие $B$ – выигрыш, т.е. сумма очков больше $10$

Здесь удобно воспользоваться **симметрией**: заменив на каждой кости число очков $x_i$ на $7 - x_i$, получим взаимно однозначное соответствие между исходами с суммой $s$ и исходами с суммой $21 - s$. Суммы $s > 10$ переходят при этом в суммы $s < 11$, т.е. эти две группы равномощны и вместе дают все $216$ исходов:

$P(B) = \frac{1}{2} = 0{,}5$

Тот же результат получается прямым суммированием числа исходов для сумм от $11$ до $18$.

```python
from itertools import product

dice = list(product(range(1, 7), repeat=3))

m2 = sum(1 for t in dice if sum(t) > 10)
print(f"благоприятствующих исходов: {m2} из {len(dice)}")
round(m2 / len(dice), 4)
```

**Распределение числа исходов по суммам** – наглядно видно симметрию относительно значения $10{,}5$.

```python
from collections import Counter

counts = Counter(sum(t) for t in dice)
for s in sorted(counts):
    print(f"сумма {s:>2}: {counts[s]:>2} исходов")
```
